## Day 2: Answers

**1. How many different crops are there?**
There are 22 crops (rice, maize, chickpea, kidneybeans, pigeonpeas,
mothbeans, mungbean, blackgram, lentil, pomegranate, banana, mango,
grapes, watermelon, muskmelon, apple, orange, papaya, coconut,
cotton, jute, coffee).

**2. Are they balanced?**
Yes. Every crop has exactly 100 rows (22 x 100 = 2200). This means
accuracy is a fair metric here, and random guessing would be right
only about 1 time in 22 (roughly 4.5%).

**3. Does anything in describe() look odd?**
- No missing values: every column has a count of 2200.
- pH ranges from 3.5 to 9.9. That is within the valid 0-14 scale,
  but both ends are quite extreme for soil, so worth checking later.
- Humidity ranges from 14.3 to 100, which is within 0-100.
- Potassium (K): 75% of rows are 49 or below, but the maximum is
  205. A few rows have very high K. I will check by crop before
  deciding if they are errors.
- Humidity: the median (80.5) is higher than the mean (71.5), so
  some low-humidity rows pull the average down.
- The columns are on very different scales (K up to 205, pH up to
  about 10), which matters for some models later.
